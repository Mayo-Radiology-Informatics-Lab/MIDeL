## **Generative Adversarial Networks (GANs)**

## Overview

Generative Adversarial Networks (GANs) are a class of deep learning models introduced by Goodfellow et al. (2014) that can generate realistic synthetic data. In medical imaging, GANs have found applications in:

- **Data augmentation** — generating synthetic training images to address dataset scarcity
- **Image-to-image translation** — converting between imaging modalities (e.g., MRI → CT)
- **Super-resolution** — enhancing image quality and resolution
- **Anomaly detection** — identifying pathological regions
- **De-noising** — improving image quality from low-dose acquisitions

A GAN consists of two competing networks:
- **Generator (G):** learns to produce realistic synthetic images from random noise
- **Discriminator (D):** learns to distinguish real images from generated ones

The two networks are trained simultaneously in a minimax game:

$$\min_G \max_D \; \mathbb{E}[\log D(x)] + \mathbb{E}[\log(1 - D(G(z)))]$$
