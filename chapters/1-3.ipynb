### Section 1: Introduction to Biomedical Ethics

<u>"People don't care how much you know until they know how much you care."</u> In the age of artificial intelligence, and the rapid digitization of medicine, this quote by US President Theodore Roosevelt has become of increasing importance regarding patient care. Evidence of this is seen in the fact that a report by the Pew Research center found that <u>60% of American adults are uncomfortable with the use of AI in medicine<sup>1</sup></u>. One contributing factor this study found towards these views is that 57% of survey respondents believed that the use of AI in healthcare would negatively impact their relationship with their healthcare provider. Concerns about the bias embedded in AI models used in patient care also cause hesitation regarding AI's adoption in healthcare among American patients. Consequently, *even the world’s most effective, innovative, and accurate AI models will have limited impact on healthcare if the patients that stand to benefit from it do not trust its utilization*, which underscores the importance of embedding biomedical ethics principles into every step of the medical AI development process. 

This chapter will give a brief overview of the four principles of biomedical ethics, first introduced by Childress and Beauchamp, and how they can be applied to medical AI research to develop responsible and trustworthy models<sup>2</sup>. These principles are Autonomy, beneficence, non-maleficence, and justice. These principles’ implications on medical AI research, and how responsible data scientists should incorporate them into their research protocols are detailed below. This chapter will also detail guidelines presented in the Belmont Report, which was published by the Department of Health, Education, and Welfare in 1979 to provide AI researchers with a practical set of guidelines towards the protection of research subjects in this fast-moving field<sup>3</sup>. 

### Section 2: Patient Autonomy & AI

##### <u>2.1 Definition of Autonomy & Informed Consent</u>
A simple, working definition of autonomy in biomedical ethics is that “autonomy means self-government"<sup>4</sup>. Essentially, <u>patients have both the ability and the right to make their own decisions regarding what practices and behaviors they will or will not accept regarding their own healthcare</u>. Informed consent is a widely accepted component of medical practice that, when properly employed, fulfills the medical AI researcher’s obligation to patient autonomy.  

The most widely-employed method of obtaining informed consent is via an informed-consent form, which in biomedical research is developed in collaboration with an institution’s medical ethics committee, often referred to as the Institutional Review Board (IRB). The exact contents of the informed consent form will vary depending on the potential risks and benefits subjects will be exposed to, and the policies of the institution where the research will take place. A general template of the informed consent form is available through the National Institute on Aging. A screenshot of the first page of this template, filled in with a general example of a generic retrospective study using AI to study radiological data is featured in Figure 1<sup>5</sup>.

##### <u>2.2 Common Concerns Regarding Autonomy & AI</u>
One concern regarding research subject autonomy in all medical research includes informed consent to research among vulnerable populations. Examples of such subject populations can include pregnant people, children, those of diminished mental capacity, and those with disabilities. In these populations, the primary concern of autonomy is to avoid coercion and undue influence, assist comprehension when needed, and provide these populations “the opportunity to choose to the extent they are able, whether or not to participate in research”<sup>3</sup>. Often, respect of autonomy for vulnerable populations, for example those of mental immaturity such as young children, involves “seeking the permission of [these] parties in order to protect the subject from harm”<sup>3</sup>. It is critical when developing the informed consent form to work with the relevant IRB to properly address autonomy for your research project, in these populations.  

Specific to AI research, a common concern regarding the autonomy of research subjects is the inherent opacity of AI models. It is important when crafting the informed consent process for an AI research project to provide prospective subjects with a high-level overview of the way a model will handle their data. For example, if a subject’s data will be used for model training, this should be disclosed to the subject, along with a brief, laymen’s explanation of what the training process will look like, with respect to the model’s handling of that subject’s data. Researchers should also treat the process of autonomy and obtaining informed consent as exactly that: a process wherein participants both before and after signing the informed consent form, can ask questions and receive answers. 

### Section 3: Beneficence & AI 

##### <u>3.1 Defining Beneficence </u>
As detailed in the Belmont Report, the principle of beneficence in biomedical research requires that a research team makes “efforts to secure [research subjects’] well-being"<sup>3</sup>. The Report outlines three specific components of an ethical biomedical research study:  
1. <u>“Do no harm”</u>. An inherent component of beneficence in biomedical research includes doing no harm to the human subjects involved. The concept of ‘harm’ in AI research is admittedly nuanced and subjective, in no small part because the field has so recently developed. One common cause of potential harm regarding AI research is the privacy of subjects’ data, with 53% of Americans believing that AI will “hurt more than it helps” at keeping personal information private, according to a 2023 study conducted by the Pew Research Center<sup>6</sup>. Further, in a 2024 study by the same group, 43% of American adults believed that AI will more likely harm them personally, while only 24% indicated that they believed AI will more likely benefit them personally<sup>7</sup>. As a result, data scientists must climb an uphill battle to prove that they are developing AI models in a manner that is trustworthy and handles subjects’ data with respect for their privacy. More detailed guidelines related to this concept are covered in Section 4 of this chapter. 

2. “Maximize possible benefits”. The core component of beneficence includes maximizing the benefit a research project has to the participating subjects, as well as society at large, stands to gain from participation in the research. Due to AI’s susceptibility to bias in the training dataset, one important component of maximizing benefits to individual patients, and society at large, is to include as diverse a dataset as possible into the training, fine-tuning, and final testing sets. Definitions of a “diverse” dataset for AI is ambiguous, and can include accounting for variability in a subject population’s: 
    * Racial, ethnic, and religious backgrounds 
    * Sex assigned at birth, and gender identity 
    * Age  
    * Socioeconomic status 
    * Living situation 
    * Data collection instruments and protocols (i.e., data collected from scanners made by different manufacturers, and/or scanners operated by different institutions, etc) 
    * Data processing protocols 
    * ...and many others! 

3. “Minimize possible harms”. This concept is intertwined with section 4 (Non-Maleficence and AI), and regarding research protocols specific to AI research, one of the most important risks/harms to protect against is a breech of research subject confidentiality. One of the most straightforward definitions of confidentiality is the accidental release of one or more of a subject’s 18 Health Insurance Portability and Accountability Act (HIPAA) identifiers (see list below)<sup>8</sup>. Unless absolutely necessary for the research project, AI models should not take as input any HIPAA identifiers. Additionally, an AI model should never return any of these identifiers as an output. 
    * Name 
    * Geographic location (excepting a zip code, if the population contains more than 20,000 residents) 
    * All date information (excepting year) related to the patient, including but not limited to: Birth date, date of death, admission date, and discharge date 
    * Telephone numbers 
    * Fax numbers 
    * E-mail addresses 
    * Social security numbers 
    * Medical record numbers 
    * Health plan beneficiary numbers 
    * Account numbers 
    * Certificate/license numbers 
    * Vehicle identifiers 
    * Device identifiers 
    * Web addresses 
    * Internet protocol (IP) addresses 
    * Biometric identifiers 
    * Full face photographs 
    * Any other identifying number, characteristic, or code 

##### <u>3.2 Action Items for Respecting Beneficence in AI </u>
While not an all-inclusive list, the following action items, upon completion, will lay the foundation for an ethically designed and conducted medical AI research study, with respect to beneficence: 
* Do not include HIPAA identifiers as input to the model 
* Do not include HIPAA identifiers as a target output of the model
* Ensure that the training dataset of an AI model is balanced with respect to the subjects’ demographics, including their: 
    * Sex assigned at birth 
    * Gender identity 
    * Race 
    * Ethnicity 
    * Age 
    * Socioeconomic status 
* Ensure that the training and validation of an AI model has included diversity with respect to data collection protocols, such as: 
    * Data collection instruments (i.e., scanners from different manufacturers)  
    * Data collection protocols (i.e., use of different contrast agents, data collected by different individuals, etc) 

### Section 4: Non-Maleficence & AI 

The ethical principle of non-maleficence “do no harm” is central to healthcare and must guide the development and deployment of AI systems. Patients often fear that AI could compromise their safety through breaches of confidentiality, inaccurate or biased recommendations, or the erosion of human judgment in clinical care. These concerns are well-founded: past failures in protecting patient data and the misuse of unvalidated technologies illustrate how quickly trust can be lost when safety is not prioritized. 

To uphold non-maleficence, AI scientists and healthcare professionals must take a comprehensive approach that safeguards patients at every stage: 
* **Protection of confidentiality**: Ensure that sensitive patient information is secured through encryption, anonymization, and strict data access controls. Preserving privacy is fundamental to trust in AI. 
* **Multidisciplinary collaboration**: Involve clinicians, data scientists, ethicists, and regulatory experts in development. Multidisciplinary teams help build more robust and clinically accurate models by combining technical and medical expertise. 
* **Explainability**: Favor interpretable models and techniques that allow clinicians to understand and communicate how AI generates outputs. Patients and doctors must not be left in the dark about algorithmic reasoning. 
* **Balance with human judgment:** AI should serve as a decision aid, not a replacement. Maintaining human oversight ensures that medical intuition, empathy, and clinical expertise remain central in patient care. 
* **Clinician understanding of AI:** Provide clinicians with the tools and training to critically evaluate AI outputs. Without basic AI literacy, clinicians cannot effectively integrate algorithmic insights into safe decision-making. 
* **Continuous clinical validation:** Treat validation not as a one-time hurdle but as an ongoing process. 
    * *Pre-clinical development:* Use synthetic data to stress-test algorithms in controlled conditions. 
    * *Clinical studies:* Evaluate performance on real patient data before clinical deployment. 
    * *Getting to market:* Secure regulatory approval (e.g., FDA, EMA) to ensure safety and compliance. 
    * *Post-market surveillance:* Monitor real-world use for errors, biases, or unintended harms after deployment. 
* Feedback loops from clinicians: Create mechanisms for continuous feedback from frontline healthcare professionals, allowing models to improve and adapt safely in practice. 

By integrating these practices, AI development aligns with the medical imperative to avoid harm. Protecting confidentiality, ensuring explainability, preserving human judgment, and validating continuously across the AI lifecycle are not optional add-ons, they are ethical requirements. 

### Section 5: Justice & AI 

The principle of *justice* requires that healthcare be fair, equitable, and inclusive. In the context of AI, this principle means that algorithms must not reproduce existing disparities or create new ones. Patients often worry that AI systems may disadvantage certain groups, especially when those groups are underrepresented in training datasets. These concerns are supported by real-world evidence. For example, dermatology tools developed with predominantly lighter skin images performed less accurately on darker skin tones, showing how data bias can lead to unequal outcomes. 

To uphold justice in AI, several priorities should guide both design and deployment: 

* **Bias in models** Algorithms should be assessed for bias at every stage of development and use. Mitigation strategies include balanced sampling, fairness-aware training, and subgroup performance audits. 
    * *Pre-development:* Identify and address data gaps before training begins. 
    * *Model training:* Apply bias detection and correction techniques during learning. 
    * *Post-deployment:* Audit models periodically in real-world settings to capture new sources of inequity. 
* **Transparency of models** Trust depends on openness about how algorithms are created and validated. 
    * *Patients:* Should receive clear, understandable explanations of how AI contributes to their care. 
    * *Doctors:* Need access to detailed information about model strengths, weaknesses, and subgroup performance. 
    * *Scientists:* Must disclose data sources, design decisions, and known limitations in performance. 
* **Fair and representative datasets**  Inclusive data collection is the foundation of equitable AI. 
    * *Data gathering:* Ensure representation across age, sex, race, geography, and disease subtypes. 
    * *Validation:* Test models on diverse populations to confirm reliability. 
    * *Continuous updating:* Incorporate new and broader data over time to prevent systematic exclusion. 
* **Collaboration with stakeholders** Engaging both clinicians and communities in AI design and evaluation strengthens justice. Input from patients, particularly those from historically underserved groups, ensures that systems reflect real-world needs and values. 
* **Ongoing fairness assessments** Equity must be safeguarded beyond initial deployment. 
    * *Pre-deployment testing:* Simulate different demographic scenarios to anticipate unequal outcomes. 
    * *Clinical monitoring:* Track how algorithms perform in day-to-day use across subgroups. 
    * *Corrective action:* Update or retrain models promptly if disparities emerge. 

By embedding justice into every stage of the AI lifecycle through bias mitigation, transparency, inclusive data practices, stakeholder collaboration, and continuous monitoring, healthcare can ensure that AI narrows disparities rather than widening them. Fairness in AI is a shared responsibility among patients, doctors, and scientists. 

### Section 6: Guidelines & Conclusion 

The ethical integration of AI into healthcare must be guided by the long-standing principles of biomedical ethics. Patient autonomy reminds us that individuals have the right to understand and consent to the role of AI in their care. Beneficence compels us to use these tools for the patient’s good, ensuring they meaningfully enhance health outcomes. Non-maleficence underscores the responsibility to minimize harm through confidentiality, rigorous validation, and human oversight, while justice demands equity, fairness, and inclusivity in design and deployment. Together, these principles provide a framework for aligning technological innovation with patient-centered values. 

**Key ethical principles for responsible AI in healthcare:** 
* Protect confidentiality through encryption, anonymization, and informed consent. 
* Keep human oversight at the center of clinical decision-making. 
* Treat validation as a continuous process across pre-clinical, clinical, and post-market phases. 
* Audit for bias regularly, and correct inequities in data and outputs. 
* Prioritize transparency and explainability to maintain trust. 
* Engage multidisciplinary teams and communities in development and deployment. 

By embedding these principles into every stage of development and deployment, AI becomes not only a technological advance but a trustworthy partner in patient care.
